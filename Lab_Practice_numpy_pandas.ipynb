# Task 9: Apply a Condition
Suppose grace marks connot make a student's marks exceed 100.
Create an updated marks array where:
* if marks+5>100, keep th mark as 100.
* Otherwise, add5


```python
upmarks = marks+5

for i in range (0,len(upmarks)):
    for j in range (0,len(upmarks[0,:])):
        if upmarks[i,j] > 100 :
            upmarks[i,j] = 100

print(upmarks)
```

# Task 10: Pass/Fail Analysis
A student is considered passed if their average marks are 60 or above.

Find:
* Average marks of every student
* Which students passed
* Which students failed
* Number of students who passed
* Number of students who failed


```python
avg_marks = marks.mean(axis = 1)
print(avg_marks)
a = list(avg_marks)
b= 0
c = []
d = []
for i in a:
    if i > 60 :
        c.append(a.index(i,b)+1)
    else:
        d.append(a.index(i,b)+1)
    b+= 1
print(f"Passed students are students{c}")
print(f"Number of passed students is {len(c)}")
print(f"Failed students are students{d}")
print(f"Number of failed students is {len(d)}")
```
